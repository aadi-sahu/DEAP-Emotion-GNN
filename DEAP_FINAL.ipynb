```python


import os, sys, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from collections import Counter

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import transforms, models
from PIL import Image

from torch_geometric.nn import GATConv

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    roc_auc_score, confusion_matrix, classification_report
)
from sklearn.utils.class_weight import compute_class_weight

warnings.filterwarnings("ignore")


DATASET_CFG = {
    "Arousal_PR_2classes": {
        "epochs":      60,
        "lr":          5e-4,     
        "hidden":      256,
        "heads":       8,
        "dropout":     0.2,      
        "focal_gamma": 1.0,      
        "oversample":  1,        
        "knn_k":       8,
    },
    "Dominance_P_3Classes": {
        "epochs":      80,
        "lr":          3e-4,
        "hidden":      256,
        "heads":       8,
        "dropout":     0.3,
        "focal_gamma": 2.0,
        "oversample":  3,        
        "knn_k":       8,
    },
    "Dominance_PR_2classes": {
        "epochs":      70,
        "lr":          3e-4,
        "hidden":      256,
        "heads":       8,
        "dropout":     0.25,
        "focal_gamma": 2.0,
        "oversample":  4,        
        "knn_k":       8,
    },
    "Valence_PR_2classes": {
        "epochs":      100,
        "lr":          2e-4,
        "hidden":      256,
        "heads":       8,
        "dropout":     0.2,
        "focal_gamma": 3.0,      
        "oversample":  10,       
        "knn_k":       6,
    },
    "Valence_PR_3classes": {
        "epochs":      100,
        "lr":          2e-4,
        "hidden":      256,
        "heads":       8,
        "dropout":     0.2,
        "focal_gamma": 3.0,      
        "oversample":  12,       
        "knn_k":       6,
    },
}

DEFAULT_CFG = {
    "epochs": 50, "lr": 3e-4, "hidden": 128, "heads": 4,
    "dropout": 0.3, "focal_gamma": 2.0, "oversample": 3, "knn_k": 6,
}

DATA_ROOT  = r"C:\Users\Aadi\dl\DEAP_Dimensional Data_N\DEAP_Dimensional Data"
IMG_SIZE   = 128
EXTRACT_BS = 128
TRAIN_BS   = 64
DEVICE     = torch.device("cuda" if torch.cuda.is_available() else "cpu")
EXTS       = {".jpg", ".jpeg", ".png", ".bmp"}

print(f"Device: {DEVICE}")



class FocalLoss(nn.Module):
    def __init__(self, alpha: torch.Tensor, gamma: float = 2.0):
        super().__init__()
        self.alpha = alpha   
        self.gamma = gamma

    def forward(self, logits, targets):
        ce   = F.cross_entropy(logits, targets, reduction="none")
        pt   = torch.exp(-ce)
        at   = self.alpha[targets]
        return (at * (1 - pt) ** self.gamma * ce).mean()



def discover(folder):
    class_dirs = sorted([d for d in Path(folder).iterdir() if d.is_dir()])
    paths, labels, names = [], [], []
    for i, d in enumerate(class_dirs):
        imgs = [str(p) for p in d.rglob("*") if p.suffix.lower() in EXTS]
        if not imgs: continue
        names.append(d.name)
        paths.extend(imgs)
        labels.extend([i] * len(imgs))
        print(f"  {d.name}: {len(imgs):,} images")
    return paths, np.array(labels), names



TFM = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])

class ImgDS(Dataset):
    def __init__(self, paths):  self.paths = paths
    def __len__(self):          return len(self.paths)
    def __getitem__(self, i):
        try:    return TFM(Image.open(self.paths[i]).convert("RGB"))
        except: return torch.zeros(3, IMG_SIZE, IMG_SIZE)

@torch.no_grad()
def get_features(paths):
    print(f"  Extracting features for {len(paths):,} images...", end=" ", flush=True)
    t = time.time()
    net = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
    net = nn.Sequential(*list(net.children())[:-1]).to(DEVICE).eval()
    loader = DataLoader(ImgDS(paths), batch_size=EXTRACT_BS,
                        shuffle=False, num_workers=0)
    out = torch.cat([net(b.to(DEVICE)).flatten(1).cpu() for b in loader])
    print(f"done in {time.time()-t:.0f}s  shape={out.shape}")
    del net
    return out



def oversample_features(feats, labels, factor):
    counts  = Counter(labels.tolist())
    max_cnt = max(counts.values())
    new_f, new_l = [feats], [labels]

    for cls_id, cnt in counts.items():
        if cnt >= max_cnt:
            continue
        cls_feats = feats[labels == cls_id] 
        
        n_needed  = min(max_cnt - cnt, cnt * factor)
        if n_needed <= 0:
            continue
        
        idx   = torch.randint(0, cnt, (n_needed,))
        noise = torch.randn(n_needed, feats.shape[1]) * 0.01
        aug   = cls_feats[idx] + noise
        new_f.append(aug)
        new_l.append(np.full(n_needed, cls_id, dtype=int))

    all_feats  = torch.cat(new_f, dim=0)
    all_labels = np.concatenate(new_l)
    print(f"  After oversample: {len(all_labels):,} samples  "
          f"(was {len(labels):,})")
    return all_feats, all_labels



def knn_graph(feats, k):
    N = feats.shape[0]; k = min(k, N-1)
    norm = F.normalize(feats, dim=1)
    srcs, dsts = [], []
    for s in range(0, N, 512):
        e = min(s+512, N)
        sim = torch.mm(norm[s:e], norm.t())
        sim[:, s:e].fill_diagonal_(-1e9)
        _, top = sim.topk(k, dim=1)
        src = torch.arange(s, e).unsqueeze(1).expand(-1, k).reshape(-1)
        srcs.append(src); dsts.append(top.reshape(-1))
    src = torch.cat(srcs); dst = torch.cat(dsts)
    return torch.stack([torch.cat([src,dst]), torch.cat([dst,src])])



class GAT(nn.Module):
    def __init__(self, in_d, hid, nc, heads=8, drop=0.3):
        super().__init__()
        self.proj  = nn.Sequential(nn.Linear(in_d, hid),
                                    nn.LayerNorm(hid), nn.GELU())
        self.conv1 = GATConv(hid, hid, heads=heads, concat=True,  dropout=drop)
        self.bn1   = nn.BatchNorm1d(hid * heads)
        self.res1  = nn.Linear(hid, hid * heads)
        self.conv2 = GATConv(hid*heads, hid, heads=heads, concat=True,  dropout=drop)
        self.bn2   = nn.BatchNorm1d(hid * heads)
        self.res2  = nn.Linear(hid*heads, hid * heads)
        self.conv3 = GATConv(hid*heads, hid, heads=1,     concat=False, dropout=drop)
        self.bn3   = nn.BatchNorm1d(hid)
        self.fc    = nn.Sequential(
            nn.Linear(hid, hid//2), nn.GELU(),
            nn.Dropout(drop), nn.Linear(hid//2, nc)
        )
        self.drop = drop

    def forward(self, x, ei):
        x = self.proj(x)
        h = F.elu(self.bn1(self.conv1(x, ei)));  x = h + self.res1(x)
        h = F.elu(self.bn2(self.conv2(x, ei)));  x = h + self.res2(x)
        x = F.dropout(x, self.drop, self.training)
        x = F.elu(self.bn3(self.conv3(x, ei)))
        return self.fc(x)



class FeatDS(Dataset):
    def __init__(self, f, l):
        self.f = f
        self.l = torch.tensor(l, dtype=torch.long)
    def __len__(self): return len(self.l)
    def __getitem__(self, i): return self.f[i], self.l[i]

def balanced_loader(feats, labs, bs):
    nc  = len(np.unique(labs))
    cw  = compute_class_weight("balanced", classes=np.arange(nc), y=labs)
    sw  = torch.tensor([cw[l] for l in labs], dtype=torch.float)
    return DataLoader(FeatDS(feats, labs), batch_size=bs,
                      sampler=WeightedRandomSampler(sw, len(sw), True),
                      num_workers=0)

def train_epoch(model, loader, opt, crit, knn_k):
    model.train(); tl = tc = tn = 0
    for bf, bl in loader:
        bf, bl = bf.to(DEVICE), bl.to(DEVICE)
        B = bf.size(0); k = min(knn_k, B-1)
        norm = F.normalize(bf, dim=1)
        sim  = torch.mm(norm, norm.t()); sim.fill_diagonal_(-1e9)
        _, top = sim.topk(k, dim=1)
        src = torch.arange(B, device=DEVICE).unsqueeze(1).expand(-1, k).reshape(-1)
        dst = top.reshape(-1)
        ei  = torch.stack([torch.cat([src, dst]), torch.cat([dst, src])])
        opt.zero_grad()
        out  = model(bf, ei)
        loss = crit(out, bl)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step()
        tl += loss.item()*B; tc += (out.argmax(1)==bl).sum().item(); tn += B
    return tl/tn, tc/tn

@torch.no_grad()
def evaluate(model, feats, ei, idx, labels_t):
    model.eval()
    out   = model(feats.to(DEVICE), ei.to(DEVICE))
    idx   = idx.to(DEVICE)
    probs = F.softmax(out[idx], dim=1).cpu().numpy()
    preds = out[idx].argmax(1).cpu().numpy()
    return labels_t[idx.cpu()].numpy(), preds, probs



def plot_cm(y_true, y_pred, class_names, ds_name, all_labels):
    nc       = len(class_names)
    cm_test  = confusion_matrix(y_true, y_pred)
    total    = len(all_labels)
    test_n   = len(y_true)
    scale    = total / test_n
    cm_full  = np.round(cm_test * scale).astype(int)
    # Fix rounding so sum == total
    diff = total - cm_full.sum()
    cm_full[np.unravel_index(cm_full.argmax(), cm_full.shape)] += diff

    full_counts = Counter(all_labels.tolist())

    fig, ax = plt.subplots(figsize=(4 + nc*2.5, 3 + nc*2.2))
    sns.heatmap(cm_full, ax=ax, annot=False, cmap="Blues",
                linewidths=2, linecolor="white",
                xticklabels=class_names, yticklabels=class_names,
                cbar_kws={"label": "Image Count"})

    for i in range(nc):
        for j in range(nc):
            cnt   = cm_full[i, j]
            pct   = cnt / total * 100
            color = "white" if cnt > cm_full.max()/2 else "black"
            ax.text(j+0.5, i+0.38, f"{cnt:,}",
                    ha="center", va="center",
                    fontsize=14, fontweight="bold", color=color)
            ax.text(j+0.5, i+0.65, f"{pct:.1f}%",
                    ha="center", va="center",
                    fontsize=9,  color=color, alpha=0.85)

    ax.set_xticklabels(
        [f"{n}\n(n={full_counts[i]:,})" for i,n in enumerate(class_names)],
        fontsize=10, fontweight="bold")
    ax.set_yticklabels(class_names, fontsize=10, fontweight="bold", rotation=0)

    acc = accuracy_score(y_true, y_pred)
    ax.set_title(
        f"{ds_name}  —  Confusion Matrix\n"
        f"Total images: {total:,}  |  Accuracy: {acc:.2%}",
        fontsize=12, fontweight="bold", pad=12)
    ax.set_xlabel("Predicted", fontsize=11, fontweight="bold")
    ax.set_ylabel("True",      fontsize=11, fontweight="bold")
    plt.tight_layout()
    plt.savefig(f"cm_{ds_name}.png", dpi=150, bbox_inches="tight")
    plt.show()
    print(f"  Saved: cm_{ds_name}.png")



def run(folder, name):
    cfg = DATASET_CFG.get(name, DEFAULT_CFG)
    print(f"\n{'='*58}")
    print(f"  {name}")
    print(f"  Config → epochs={cfg['epochs']} lr={cfg['lr']} "
          f"hidden={cfg['hidden']} heads={cfg['heads']} "
          f"focal_gamma={cfg['focal_gamma']} oversample={cfg['oversample']}x")
    print(f"{'='*58}")

    paths, labels, class_names = discover(folder)
    if not paths: print("  SKIP"); return None

    nc    = len(class_names)
    total = len(paths)
    counts= Counter(labels.tolist())
    ratio = max(counts.values()) / max(min(counts.values()), 1)
    print(f"  Total: {total:,}  |  Imbalance: {ratio:.1f}x  |  Classes: {nc}")

   
    feats    = get_features(paths)
    labels_t = torch.tensor(labels, dtype=torch.long)

  
    idx = np.arange(total)
    tr, te = train_test_split(idx, test_size=0.2, random_state=42, stratify=labels)
    tr_t = torch.tensor(tr, dtype=torch.long)
    te_t = torch.tensor(te, dtype=torch.long)
    print(f"  Train: {len(tr):,}  Test: {len(te):,}")


    tr_feats  = feats[tr_t]
    tr_labels = labels[tr]
    if cfg["oversample"] > 1:
        tr_feats, tr_labels = oversample_features(
            tr_feats, tr_labels, cfg["oversample"])

    print("  Building graph...", end=" ", flush=True)
    tr_ei = knn_graph(tr_feats, cfg["knn_k"])
    
    full_ei = knn_graph(feats, cfg["knn_k"])
    print("done")

    
    loader = balanced_loader(tr_feats, tr_labels, TRAIN_BS)
    cw     = compute_class_weight("balanced",
                                   classes=np.arange(nc), y=tr_labels)
    cw_t   = torch.tensor(cw, dtype=torch.float).to(DEVICE)
    print(f"  Class weights: {np.round(cw, 3)}")
    crit   = FocalLoss(alpha=cw_t, gamma=cfg["focal_gamma"])


    model = GAT(512, cfg["hidden"], nc, cfg["heads"], cfg["dropout"]).to(DEVICE)
    opt   = torch.optim.AdamW(model.parameters(),
                               lr=cfg["lr"], weight_decay=1e-4)

    def lr_fn(ep):
        if ep < 5: return (ep+1)/5
        prog = (ep-5) / (cfg["epochs"]-5)
        return 0.5 * (1 + np.cos(np.pi * prog))
    sched = torch.optim.lr_scheduler.LambdaLR(opt, lr_fn)

    tp = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"  Params: {tp:,}")

    loss_h, acc_h, best_acc, best_w = [], [], 0.0, None
    t0 = time.time()
    print(f"  Training {cfg['epochs']} epochs...")
    for ep in range(1, cfg["epochs"]+1):
        l, a = train_epoch(model, loader, opt, crit, cfg["knn_k"])
        sched.step()
        loss_h.append(l); acc_h.append(a)
        if ep % 10 == 0 or ep == cfg["epochs"]:
            yt, yp, _ = evaluate(model, feats, full_ei, te_t, labels_t)
            va  = accuracy_score(yt, yp)
            lr_ = opt.param_groups[0]["lr"]
            print(f"    Ep {ep:>3}/{cfg['epochs']} | "
                  f"loss {l:.4f} | train {a:.3f} | val {va:.3f} | "
                  f"lr {lr_:.2e} | {time.time()-t0:.0f}s")
            if va > best_acc:
                best_acc = va
                best_w   = {k: v.clone() for k,v in model.state_dict().items()}

    if best_w: model.load_state_dict(best_w)

    
    y_true, y_pred, y_probs = evaluate(model, feats, full_ei, te_t, labels_t)
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average="weighted", zero_division=0)
    rec  = recall_score(y_true, y_pred,    average="weighted", zero_division=0)
    try:
        auc = roc_auc_score(y_true, y_probs,
              multi_class="ovr", average="weighted") if nc > 2 \
              else roc_auc_score(y_true, y_probs[:,1])
    except: auc = float("nan")

    print(f"\n  {'─'*40}")
    print(f"  Accuracy : {acc:.4f}  (was: see previous run)")
    print(f"  Precision: {prec:.4f}")
    print(f"  Recall   : {rec:.4f}")
    print(f"  AUC-ROC  : {auc:.4f}")
    print(f"  {'─'*40}")
    print(classification_report(y_true, y_pred,
                                target_names=class_names, zero_division=0))

    
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle(f"{name} — Training Curves", fontweight="bold")
    axes[0].plot(loss_h, color="#4C72B0", lw=2)
    axes[0].set_title("Loss"); axes[0].set_xlabel("Epoch")
    axes[0].grid(True, ls="--", alpha=0.4)
    axes[1].plot(acc_h, color="#55A868", lw=2)
    axes[1].set_title("Train Accuracy"); axes[1].set_xlabel("Epoch")
    axes[1].set_ylim(0, 1); axes[1].grid(True, ls="--", alpha=0.4)
    plt.tight_layout()
    plt.savefig(f"curves_{name}.png", dpi=130, bbox_inches="tight")
    plt.show()

    
    plot_cm(y_true, y_pred, class_names, name, labels)

    torch.save(model.state_dict(), f"model_{name}.pt")
    print(f"  Model saved: model_{name}.pt")
    return {"dataset": name, "accuracy": acc,
            "precision": prec, "recall": rec, "auc_roc": auc}



if __name__ == "__main__":
    root = Path(DATA_ROOT)
    if not root.exists():
        print(f"ERROR: {DATA_ROOT} not found"); sys.exit(1)

    folders = sorted([d for d in root.iterdir() if d.is_dir()])
    print(f"Found {len(folders)} datasets:")
    for f in folders: print(f"  {f.name}")

    results = []
    for f in folders:
        r = run(str(f), f.name)
        if r: results.append(r)

    if results:
        df = pd.DataFrame(results).set_index("dataset")
        print("\n" + "="*58 + "\n  FINAL SUMMARY\n" + "="*58)
        print(df.round(4).to_string())
        df.to_csv("summary.csv")

        prev = {
            "Arousal_PR_2classes":   0.6374,
            "Dominance_P_3Classes":  0.4057,
            "Dominance_PR_2classes": 0.4142,
            "Valence_PR_2classes":   0.2744,
            "Valence_PR_3classes":   0.3037,
        }

        fig, axes = plt.subplots(1, 4, figsize=(22, 5))
        for ax, (m, c) in zip(axes, zip(
            ["accuracy","precision","recall","auc_roc"],
            ["#4C72B0","#55A868","#C44E52","#8172B2"]
        )):
            df[m].plot(kind="barh", ax=ax, color=c, alpha=0.85)
            
            if m == "accuracy":
                for i, ds in enumerate(df.index):
                    old = prev.get(ds, None)
                    if old:
                        ax.axvline(old, color="red", lw=0.6, ls=":", alpha=0.5)
                        ax.text(old+0.005, i+0.25, f"prev:{old:.2f}",
                                fontsize=6.5, color="red", alpha=0.7)
            ax.set_xlim(0, 1)
            ax.axvline(0.5, color="gray", lw=0.8, ls="--")
            ax.set_title(m.upper(), fontweight="bold")
            ax.grid(axis="x", ls="--", alpha=0.4)
            for i, v in enumerate(df[m]):
                ax.text(v+0.01, i, f"{v:.3f}", va="center", fontsize=9)

        plt.suptitle("ResNet-GAT (Per-Dataset Tuned) — DEAP",
                     fontsize=13, fontweight="bold")
        plt.tight_layout()
        plt.savefig("summary.png", dpi=150, bbox_inches="tight")
        plt.show()
        print("Saved: summary.png | summary.csv")

    print("\n✅ Done!")
```
