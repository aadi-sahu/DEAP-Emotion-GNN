```python
import os, sys, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from collections import Counter

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import transforms, models
from PIL import Image

from torch_geometric.nn import GATConv

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    roc_auc_score, confusion_matrix, classification_report
)
from sklearn.utils.class_weight import compute_class_weight

warnings.filterwarnings("ignore")

DATA_ROOT  = r"C:\Users\Aadi\dl\DEAP_Dimensional Data_N\DEAP_Dimensional Data"
IMG_SIZE   = 96
EXTRACT_BS = 128
TRAIN_BS   = 64
EPOCHS     = 25
LR         = 2e-4
HIDDEN     = 128
HEADS      = 4
DROPOUT    = 0.3
KNN_K      = 6
DEVICE     = torch.device("cuda" if torch.cuda.is_available() else "cpu")
EXTS       = {".jpg", ".jpeg", ".png", ".bmp"}

print(f"Device: {DEVICE}")

LEVEL_MAP = {
    "Arousal_PR_2classes":   {"HA": "High", "LA": "Low"},
    "Dominance_PR_2classes": {"HD": "High", "LD": "Low"},
    "Dominance_P_3Classes":  {"HD": "High", "LD": "Low", "MD": "Mid"},
    "Valence_PR_2classes":   {"HV": "High", "LV": "Low"},
    "Valence_PR_3classes":   {"HV": "High", "LV": "Low", "MV": "Mid"},
}


EXPERIMENTS = [
    
    {
        "name":  "HighVsLow_AllDatasets",
        "title": "High vs Low Emotion (All Dimensions)",
        "classes": ["High", "Low"],
        "desc":  "HA+HV+HD  vs  LA+LV+LD"
    },
    {
        "name":  "HighVsMid_AllDatasets",
        "title": "High vs Mid Emotion (All Dimensions)",
        "classes": ["High", "Mid"],
        "desc":  "HA+HV+HD  vs  MV+MD"
    },
    {
        "name":  "LowVsMid_AllDatasets",
        "title": "Low vs Mid Emotion (All Dimensions)",
        "classes": ["Low", "Mid"],
        "desc":  "LA+LV+LD  vs  MV+MD"
    },
    {
        "name":  "High_vs_Low_vs_Mid_AllDatasets",
        "title": "High vs Low vs Mid — 3 Class (All Dimensions)",
        "classes": ["High", "Low", "Mid"],
        "desc":  "HA+HV+HD  vs  LA+LV+LD  vs  MV+MD"
    },

    
    {
        "name":  "CrossDim_HighArousal_vs_HighValence",
        "title": "Cross-Dim: High Arousal vs High Valence",
        "classes": ["HighArousal", "HighValence"],
        "custom": [("Arousal_PR_2classes",  "HA", "HighArousal"),
                   ("Valence_PR_2classes",  "HV", "HighValence")],
        "desc":  "HA vs HV — same level, different dimension"
    },
    {
        "name":  "CrossDim_LowArousal_vs_LowValence",
        "title": "Cross-Dim: Low Arousal vs Low Valence",
        "classes": ["LowArousal", "LowValence"],
        "custom": [("Arousal_PR_2classes", "LA", "LowArousal"),
                   ("Valence_PR_2classes", "LV", "LowValence")],
        "desc":  "LA vs LV — same level, different dimension"
    },
    {
        "name":  "CrossDim_HighArousal_vs_HighDominance",
        "title": "Cross-Dim: High Arousal vs High Dominance",
        "classes": ["HighArousal", "HighDominance"],
        "custom": [("Arousal_PR_2classes",   "HA", "HighArousal"),
                   ("Dominance_PR_2classes",  "HD", "HighDominance")],
        "desc":  "HA vs HD — same level, different dimension"
    },
    {
        "name":  "CrossDim_HighValence_vs_HighDominance",
        "title": "Cross-Dim: High Valence vs High Dominance",
        "classes": ["HighValence", "HighDominance"],
        "custom": [("Valence_PR_2classes",   "HV", "HighValence"),
                   ("Dominance_PR_2classes",  "HD", "HighDominance")],
        "desc":  "HV vs HD — same level, different dimension"
    },
    {
        "name":  "CrossDim_3way_HighAVD",
        "title": "Cross-Dim 3-way: High Arousal vs High Valence vs High Dominance",
        "classes": ["HighArousal", "HighValence", "HighDominance"],
        "custom": [("Arousal_PR_2classes",  "HA", "HighArousal"),
                   ("Valence_PR_2classes",  "HV", "HighValence"),
                   ("Dominance_PR_2classes","HD", "HighDominance")],
        "desc":  "HA vs HV vs HD — 3 dimensions, same level"
    },
    {
        "name":  "CrossDim_3way_LowAVD",
        "title": "Cross-Dim 3-way: Low Arousal vs Low Valence vs Low Dominance",
        "classes": ["LowArousal", "LowValence", "LowDominance"],
        "custom": [("Arousal_PR_2classes",  "LA", "LowArousal"),
                   ("Valence_PR_2classes",  "LV", "LowValence"),
                   ("Dominance_PR_2classes","LD", "LowDominance")],
        "desc":  "LA vs LV vs LD — 3 dimensions, same level"
    },
]


LOAD_TFM = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225]),
])

def load_folder_class(folder, subfolder):
    path = Path(DATA_ROOT) / folder / subfolder
    if not path.exists(): return None
    imgs = []
    for p in path.rglob("*"):
        if p.suffix.lower() not in EXTS: continue
        try: imgs.append(LOAD_TFM(Image.open(p).convert("RGB")))
        except: pass
    return torch.stack(imgs) if imgs else None


@torch.no_grad()
def get_features(imgs):
    net = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
    net = nn.Sequential(*list(net.children())[:-1]).to(DEVICE).eval()
    out = []
    for s in range(0, len(imgs), EXTRACT_BS):
        out.append(net(imgs[s:s+EXTRACT_BS].to(DEVICE)).flatten(1).cpu())
    del net
    return torch.cat(out)


def build_dataset(exp):
    class_feats  = {}
    class_counts = {}

    if "custom" in exp:
        for folder, subfolder, label in exp["custom"]:
            print(f"  Loading {folder}/{subfolder} → '{label}'", end=" ", flush=True)
            imgs = load_folder_class(folder, subfolder)
            if imgs is None: print("NOT FOUND"); continue
            print(f"{len(imgs):,} imgs", end=" → feats... ", flush=True)
            t = time.time()
            f = get_features(imgs); del imgs
            print(f"{time.time()-t:.0f}s")
            class_feats.setdefault(label, []).append(f)
            class_counts[label] = class_counts.get(label, 0) + len(f)
    else:
        target_levels  = set(exp["classes"])
        folders_to_use = exp.get("folders", list(LEVEL_MAP.keys()))
        for folder in folders_to_use:
            if folder not in LEVEL_MAP: continue
            for subfolder, level in LEVEL_MAP[folder].items():
                if level not in target_levels: continue
                print(f"  Loading {folder}/{subfolder} → '{level}'", end=" ", flush=True)
                imgs = load_folder_class(folder, subfolder)
                if imgs is None: print("NOT FOUND"); continue
                print(f"{len(imgs):,} imgs", end=" → feats... ", flush=True)
                t = time.time()
                f = get_features(imgs); del imgs
                print(f"{time.time()-t:.0f}s")
                class_feats.setdefault(level, []).append(f)
                class_counts[level] = class_counts.get(level, 0) + len(f)

    if len(class_feats) < 2:
        print("  [SKIP] < 2 classes found"); return None, None, None, None

    class_names = sorted(class_feats.keys())
    all_f, all_l = [], []
    for idx, name in enumerate(class_names):
        f = torch.cat(class_feats[name])
        all_f.append(f)
        all_l.append(np.full(len(f), idx, dtype=int))
        print(f"  Class '{name}': {len(f):,} total samples")

    feats  = torch.cat(all_f)
    labels = np.concatenate(all_l)
    counts = Counter(labels.tolist())
    ratio  = max(counts.values()) / max(min(counts.values()), 1)
    print(f"  Total: {len(labels):,} | Imbalance: {ratio:.1f}x")
    return feats, labels, class_names, class_counts


class FocalLoss(nn.Module):
    def __init__(self, alpha, gamma=2.0):
        super().__init__()
        self.alpha = alpha; self.gamma = gamma
    def forward(self, logits, targets):
        ce    = F.cross_entropy(logits, targets, weight=self.alpha, reduction="none")
        pt    = torch.exp(-ce)
        focal = (1 - pt) ** self.gamma * ce
        nc    = logits.size(1)
        loss  = torch.zeros(nc, device=logits.device)
        for c in range(nc):
            m = targets == c
            if m.sum() > 0: loss[c] = focal[m].mean()
        return loss.mean()


class GAT(nn.Module):
    def __init__(self, in_d, hid, nc, heads=4, drop=0.3):
        super().__init__()
        self.proj  = nn.Sequential(nn.Linear(in_d, hid),
                                    nn.LayerNorm(hid), nn.GELU())
        self.conv1 = GATConv(hid, hid, heads=heads, concat=True,  dropout=drop)
        self.bn1   = nn.BatchNorm1d(hid * heads)
        self.res1  = nn.Linear(hid, hid * heads)
        self.conv2 = GATConv(hid*heads, hid, heads=1, concat=False, dropout=drop)
        self.bn2   = nn.BatchNorm1d(hid)
        self.fc    = nn.Sequential(
            nn.Linear(hid, hid//2), nn.GELU(),
            nn.Dropout(drop), nn.Linear(hid//2, nc))
        self.drop = drop
    def forward(self, x, ei):
        x = self.proj(x)
        h = F.elu(self.bn1(self.conv1(x, ei))); x = h + self.res1(x)
        x = F.dropout(x, self.drop, self.training)
        x = F.elu(self.bn2(self.conv2(x, ei)))
        return self.fc(x)


def knn_graph(feats, k=KNN_K):
    N = feats.shape[0]; k = min(k, N-1)
    norm = F.normalize(feats, dim=1)
    srcs, dsts = [], []
    for s in range(0, N, 512):
        e = min(s+512, N)
        sim = torch.mm(norm[s:e], norm.t())
        sim[:, s:e].fill_diagonal_(-1e9)
        _, top = sim.topk(k, dim=1)
        src = torch.arange(s,e).unsqueeze(1).expand(-1,k).reshape(-1)
        srcs.append(src); dsts.append(top.reshape(-1))
    src = torch.cat(srcs); dst = torch.cat(dsts)
    return torch.stack([torch.cat([src,dst]), torch.cat([dst,src])])


class FeatDS(Dataset):
    def __init__(self, f, l):
        self.f = f; self.l = torch.tensor(l, dtype=torch.long)
    def __len__(self): return len(self.l)
    def __getitem__(self, i): return self.f[i], self.l[i]

def balanced_loader(feats, labs):
    nc  = len(np.unique(labs))
    cw  = compute_class_weight("balanced", classes=np.arange(nc), y=labs)
    sw  = torch.tensor([cw[l] for l in labs], dtype=torch.float)
    return DataLoader(FeatDS(feats, labs), batch_size=TRAIN_BS,
                      sampler=WeightedRandomSampler(sw, len(sw), True),
                      num_workers=0)

def train_epoch(model, loader, opt, crit):
    model.train(); tl = tc = tn = 0
    for bf, bl in loader:
        bf, bl = bf.to(DEVICE), bl.to(DEVICE)
        B = bf.size(0); k = min(KNN_K, B-1)
        norm = F.normalize(bf, dim=1)
        sim  = torch.mm(norm, norm.t()); sim.fill_diagonal_(-1e9)
        _, top = sim.topk(k, dim=1)
        src = torch.arange(B, device=DEVICE).unsqueeze(1).expand(-1,k).reshape(-1)
        ei  = torch.stack([torch.cat([src, top.reshape(-1)]),
                            torch.cat([top.reshape(-1), src])])
        opt.zero_grad()
        out  = model(bf, ei); loss = crit(out, bl)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step()
        tl += loss.item()*B; tc += (out.argmax(1)==bl).sum().item(); tn += B
    return tl/tn, tc/tn

@torch.no_grad()
def evaluate(model, feats, ei, idx, labels_t):
    model.eval()
    out   = model(feats.to(DEVICE), ei.to(DEVICE))
    probs = F.softmax(out[idx.to(DEVICE)], dim=1).cpu().numpy()
    preds = out[idx.to(DEVICE)].argmax(1).cpu().numpy()
    return labels_t[idx].numpy(), preds, probs


def plot_cm(y_true, y_pred, class_names, exp, class_counts):
    nc     = len(class_names)
    total  = sum(class_counts.values())
    cm     = confusion_matrix(y_true, y_pred)
    scale  = total / len(y_true)
    cm_full= np.round(cm * scale).astype(int)
    diff   = total - cm_full.sum()
    cm_full[np.unravel_index(cm_full.argmax(), cm_full.shape)] += diff

    fig, ax = plt.subplots(figsize=(4 + nc*2.8, 3 + nc*2.4))
    sns.heatmap(cm_full, ax=ax, annot=False, cmap="Blues",
                linewidths=2, linecolor="white",
                xticklabels=class_names, yticklabels=class_names,
                cbar_kws={"label": "Image Count"})
    for i in range(nc):
        for j in range(nc):
            cnt   = cm_full[i, j]
            pct   = cnt / total * 100
            color = "white" if cnt > cm_full.max()/2 else "black"
            ax.text(j+0.5, i+0.35, f"{cnt:,}",
                    ha="center", va="center",
                    fontsize=13, fontweight="bold", color=color)
            ax.text(j+0.5, i+0.62, f"{pct:.1f}%",
                    ha="center", va="center",
                    fontsize=8.5, color=color, alpha=0.85)
    ax.set_xticklabels(
        [f"{n}\n(n={class_counts.get(n,0):,})" for n in class_names],
        fontsize=9, fontweight="bold")
    ax.set_yticklabels(class_names, fontsize=9, fontweight="bold", rotation=0)
    acc = accuracy_score(y_true, y_pred)
    ax.set_title(
        f"{exp['title']}\n{exp['desc']}\n"
        f"Total: {total:,} images  |  Accuracy: {acc:.2%}",
        fontsize=10, fontweight="bold", pad=10)
    ax.set_xlabel("Predicted", fontsize=10, fontweight="bold")
    ax.set_ylabel("True",      fontsize=10, fontweight="bold")
    plt.tight_layout()
    fname = f"cm_{exp['name']}.png"
    plt.savefig(fname, dpi=150, bbox_inches="tight")
    plt.show()
    print(f"  Saved: {fname}")


def run_experiment(exp):
    print(f"\n{'='*60}")
    print(f"  {exp['title']}")
    print(f"  {exp['desc']}")
    print(f"{'='*60}")

    feats, labels, class_names, class_counts = build_dataset(exp)
    if feats is None: return None

    nc       = len(class_names)
    total    = len(labels)
    labels_t = torch.tensor(labels, dtype=torch.long)

    idx = np.arange(total)
    tr, te = train_test_split(idx, test_size=0.2, random_state=42, stratify=labels)
    tr_t = torch.tensor(tr, dtype=torch.long)
    te_t = torch.tensor(te, dtype=torch.long)
    print(f"  Train: {len(tr):,}  Test: {len(te):,}")

    print("  Building graph...", end=" ", flush=True)
    ei = knn_graph(feats); print("done")

    tr_labs = labels[tr]
    cw   = compute_class_weight("balanced", classes=np.arange(nc), y=tr_labs)
    cw_t = torch.tensor(cw, dtype=torch.float).to(DEVICE)
    crit = FocalLoss(cw_t, gamma=2.0)
    loader = balanced_loader(feats[tr_t], tr_labs)
    print(f"  Weights: {dict(zip(class_names, np.round(cw,3)))}")

    model = GAT(512, HIDDEN, nc, HEADS, DROPOUT).to(DEVICE)
    opt   = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, EPOCHS, eta_min=1e-5)

    best_acc, best_w = 0.0, None
    t0 = time.time()
    print(f"  Training {EPOCHS} epochs...")
    for ep in range(1, EPOCHS+1):
        l, ta = train_epoch(model, loader, opt, crit)
        sched.step()
        if ep % 5 == 0 or ep == EPOCHS:
            yt, yp, _ = evaluate(model, feats, ei, te_t, labels_t)
            va = accuracy_score(yt, yp)
            print(f"    Ep {ep:>2}/{EPOCHS} | loss {l:.4f} | "
                  f"train {ta:.3f} | val {va:.3f} | {time.time()-t0:.0f}s")
            if va > best_acc:
                best_acc = va
                best_w   = {k: v.clone() for k,v in model.state_dict().items()}

    model.load_state_dict(best_w)

    y_true, y_pred, y_probs = evaluate(model, feats, ei, te_t, labels_t)
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average="weighted", zero_division=0)
    rec  = recall_score(y_true, y_pred,    average="weighted", zero_division=0)
    try:
        auc = roc_auc_score(y_true, y_probs,
              multi_class="ovr", average="weighted") if nc > 2 \
              else roc_auc_score(y_true, y_probs[:,1])
    except: auc = float("nan")

    print(f"\n  Accuracy : {acc:.4f}")
    print(f"  Precision: {prec:.4f}")
    print(f"  Recall   : {rec:.4f}")
    print(f"  AUC-ROC  : {auc:.4f}")
    print(classification_report(y_true, y_pred,
                                target_names=class_names, zero_division=0))

    plot_cm(y_true, y_pred, class_names, exp, class_counts)
    torch.save(model.state_dict(), f"model_{exp['name']}.pt")

    del feats
    return {"experiment": exp["name"], "desc": exp["desc"],
            "n_classes": nc, "total": total,
            "accuracy": acc, "precision": prec,
            "recall": rec, "auc_roc": auc}


if __name__ == "__main__":
    root = Path(DATA_ROOT)
    if not root.exists():
        print(f"ERROR: {DATA_ROOT} not found"); sys.exit(1)

    print(f"Running {len(EXPERIMENTS)} experiments\n")
    results = []
    for exp in EXPERIMENTS:
        r = run_experiment(exp)
        if r: results.append(r)

    if results:
        df = pd.DataFrame(results).set_index("experiment")
        print("\n" + "="*60 + "\n  SUMMARY\n" + "="*60)
        print(df.round(4).to_string())
        df.to_csv("interclass_summary.csv")

        fig, axes = plt.subplots(1, 4, figsize=(22, max(6, len(df)*0.6)))
        for ax, (m, c) in zip(axes, zip(
            ["accuracy","precision","recall","auc_roc"],
            ["#4C72B0","#55A868","#C44E52","#8172B2"]
        )):
            vals  = df[m].values
            short = [i.replace("_"," ") for i in df.index]
            ax.barh(range(len(vals)), vals, color=c, alpha=0.85)
            ax.set_yticks(range(len(vals)))
            ax.set_yticklabels(short, fontsize=8)
            ax.set_xlim(0,1)
            ax.axvline(0.5, color="gray", lw=0.8, ls="--")
            ax.set_title(m.upper(), fontweight="bold")
            ax.grid(axis="x", ls="--", alpha=0.4)
            for i, v in enumerate(vals):
                ax.text(v+0.01, i, f"{v:.3f}", va="center", fontsize=8)
        plt.suptitle("DEAP Interclass Emotion Analysis",
                     fontsize=13, fontweight="bold")
        plt.tight_layout()
        plt.savefig("interclass_summary.png", dpi=150, bbox_inches="tight")
        plt.show()
        print("Saved: interclass_summary.png | interclass_summary.csv")

    print("\n✅ Done!")
```
